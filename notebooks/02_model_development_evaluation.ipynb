## Methodology to explain
- Target: `default payment next month` (1=default, 0=no default).
- ID excluded from predictors.
- SEX, EDUCATION and MARRIAGE are one-hot encoded; repayment-status codes remain numeric in this initial version.
- Imputers, encoders and scalers are inside the pipeline and are fit on training data only.
- Stratified 80/20 development/test split, then 90/10 train/validation split within development.
- Candidate selection uses validation default-class F1 with ROC-AUC tie-break; the test set is held out.
