# Dataset inspection and EDA
Run from the repository root or adjust paths. The actual workbook field names are on spreadsheet row 2.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA=Path('data/credit_card_default_clean.csv')
df=pd.read_csv(DATA)
TARGET='default payment next month'
print(df.shape); display(df.head()); df.info()
```

```python
print('Missing cells:',df.isna().sum().sum())
print('Duplicate rows:',df.duplicated().sum())
print('Unique IDs:',df.ID.nunique())
display(df[TARGET].value_counts().to_frame('count'))
display((df[TARGET].value_counts(normalize=True)*100).round(2).to_frame('percent'))
```

```python
df.groupby('PAY_0')[TARGET].agg(['count','sum','mean']).assign(default_rate_pct=lambda d:d['mean']*100)
```
